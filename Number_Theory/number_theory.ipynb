GCD: Largest positive integer that divides two or more numbers without leaving any remainder is called GCD(greatest common divisor).

```python
def gcd(a, b):
    while b:
        print(f"gcd({a}, {b})")
        a, b = b, a % b
    return a

print(gcd(252, 105))
```

LCM (Least Common Multiple) is the smallest positive number that is exactly divisible by two or more given numbers.
Relation between gcd and lcm is gcd(a,b)*lcm(a,b)=a*b

```python
def gcd(a, b):
    while b:
        a, b = b, a % b
    return a


def lcm(a, b):
    return (a * b) // gcd(a, b)


print("LCM =", lcm(252, 105))
```

Modular Inverse
Modular Inverse is an important concept in Number Theory and Cryptography, especially in the RSA encryption algorithm.

Let's understand it step by step.
1. What is Modular Inverse?
The modular inverse of a number a modulo m is a number x such that:

{(a*x)\mod m=1}

In simple words, modular inverse means finding a number that, when multiplied by the given number and divided by the modulus, leaves a remainder of 1.


3. When does Modular Inverse exist?
A modular inverse exists only when:
{GCD(a,m)=1}
This means a and m must be coprime.

```python
def modular_inverse(a, m):
    for x in range(1, m):
        if (a * x) % m == 1:
            return x
    return -1


print(modular_inverse(3, 7))
```

Sieve of Eratosthenes
The Sieve of Eratosthenes is an efficient algorithm used to find all prime numbers up to a given number n.

How it works:
1. Start with numbers from 2 to n.
2. Take the smallest unmarked number (prime).
3. Mark all its multiples as non-prime.
4. Repeat until you reach √n.
5. The remaining unmarked numbers are prime.

```python
def sieve(n):
    prime = [True] * (n + 1)
    prime[0] = prime[1] = False

    p = 2
    while p * p <= n:
        if prime[p]:
            for i in range(p * p, n + 1, p):
                prime[i] = False
        p += 1

    result = []
    for i in range(2, n + 1):
        if prime[i]:
            result.append(i)

    return result


print(sieve(50))
print(len(sieve(1000000)))
```

Prime Factorization
Prime Factorization means expressing a number as a product (multiplication) of its prime numbers.

In simple words: Breaking a number into prime factors.

```python
def prime_factorization(n):
    factors = []

    p = 2
    while p * p <= n:
        while n % p == 0:
            factors.append(p)
            n = n // p
        p += 1

    if n > 1:
        factors.append(n)

    return factors


print(prime_factorization(60))
```
